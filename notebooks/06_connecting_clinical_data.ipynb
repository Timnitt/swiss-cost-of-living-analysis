```python
import pandas as pd

patients = pd.read_csv('../data/raw/patients.csv')
encounters = pd.read_csv('../data/raw/encounters.csv')
conditions = pd.read_csv('../data/raw/conditions.csv')
observations = pd.read_csv('../data/raw/observations.csv')
print("Patients :", patients.shape)
print("Encounters :", encounters.shape)
print("Conditions :", conditions.shape)
print("Observations :", observations.shape)
```

```python
# identify the key to connect the tables
print(patients.columns)
print(encounters.columns)
print(conditions.columns)
print(observations.columns)
```

```python
# check if patient ID matches across tables
print(len(patients['Id'].unique()))
print(len(encounters['PATIENT'].unique()))
print(len(conditions['PATIENT'].unique()))
print(len(observations['PATIENT'].unique()))
```

```python
# check whether every encounter belongs to a patient
encounter_patient_ids = set(encounters["PATIENT"])

patient_ids = set(patients["Id"])

print(len(encounter_patient_ids - patient_ids))

conditions_patient_ids = set(conditions["PATIENT"])

print(len(conditions_patient_ids - patient_ids))
```

```python
# combine encounters with patient information

encounters_with_patients = encounters.merge(patients[["Id", "BIRTHDATE", "GENDER"]], 
                                            left_on="PATIENT", 
                                            right_on="Id", 
                                            how="left")

encounters_with_patients.head(5)
```
